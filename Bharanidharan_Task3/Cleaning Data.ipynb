## Handle missing values

```python
df_cleaned = df_original.iloc[2:].copy()

df_cleaned = df_cleaned.rename(columns={"Segment>>": "Order ID"})

df_cleaned = df_cleaned.dropna(subset=["Order ID"])

print("Missing values before:", df_original.isnull().sum().sum())
print("Missing values after:", df_cleaned.isnull().sum().sum())
```

## Remove Duplicate Rows

```python
duplicates_before = df_cleaned.duplicated().sum()

df_cleaned = df_cleaned.drop_duplicates().copy()

print("Duplicates removed:", duplicates_before)
print("Duplicates remaining:", df_cleaned.duplicated().sum())
```

## Data Standardization

```python
segments = ["Consumer", "Corporate", "Home Office"]
ship_modes = ["First Class", "Same Day", "Second Class", "Standard Class"]

sales_columns = [
    f"{segment}_{mode}"
    for segment in segments
    for mode in ship_modes
]

total_columns = [
    "Consumer Total", "Corporate Total", "Home Office Total"
]

df_cleaned = df_cleaned.drop(columns=total_columns)
df_cleaned.columns = ["Order ID"] + sales_columns

df_cleaned[sales_columns] = df_cleaned[sales_columns].apply(
    pd.to_numeric, errors="coerce"
)

display(df_cleaned.head())
```

## Outlier Detection and Treatment

```python
numeric_cols = df_cleaned.select_dtypes(include=np.number).columns
outlier_results = []

for col in numeric_cols:
    Q1 = df_cleaned[col].quantile(0.25)
    Q3 = df_cleaned[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df_cleaned[col] < lower) |
                (df_cleaned[col] > upper)).sum()

    outlier_results.append({
        "Column": col,
        "Outliers": outliers,
        "Treatment": "Retained"
    })

outlier_report = pd.DataFrame(outlier_results)

display(outlier_report)
```

### Outlier Treatment Decision

- **Method:** IQR (Interquartile Range).
- **Detection:** Identified unusually high or low values in numeric sales columns.
- **Treatment:** Retained all detected outliers.
- **Justification:** High sales values may represent genuine large transactions. Removing or capping them without verification could distort the dataset.

## Data Type Correction

```python
df_cleaned["Order ID"] = df_cleaned["Order ID"].astype("string")

for col in sales_columns:
    df_cleaned[col] = pd.to_numeric(
        df_cleaned[col], errors="coerce"
    ).astype(float)

print(df_cleaned.dtypes)
```

## Before vs After Summary

```python
summary = pd.DataFrame({
    "Metric": ["Row Count", "Null Count", "Duplicate Count", "Dtype Accuracy (%)"],
    "Before": [
        len(df_original),
        df_original.isnull().sum().sum(),
        df_original.duplicated().sum(),
        round(sum(pd.api.types.is_numeric_dtype(t) for t in df_original.dtypes[1:]) / 15 * 100, 2)
    ],
    "After": [
        len(df_cleaned),
        df_cleaned.isnull().sum().sum(),
        df_cleaned.duplicated().sum(),
        round((pd.api.types.is_string_dtype(df_cleaned["Order ID"]) +
               sum(pd.api.types.is_float_dtype(df_cleaned[c]) for c in sales_columns)) / 13 * 100, 2)
    ]
})

display(summary)
```

## Save Cleaned Dataset

```python
df_cleaned[sales_columns] = df_cleaned[sales_columns].fillna(0)

print("Remaining missing values:",
      df_cleaned.isnull().sum().sum())
```

```python
df_cleaned.to_csv("cleaned_sales_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```

## Data Cleaning Decision Documentation

### 1. Data Quality Assessment
- Decided to examine missing values, duplicates, data types, and numerical anomalies before cleaning.
- Used the original dataset as a reference to compare data quality before and after cleaning.

### 2. Missing Value Handling
- Removed unnecessary header rows and records with missing Order IDs.
- Chose to replace missing sales values with zero where blank cells represent no recorded sales, rather than using mean or median imputation.

### 3. Duplicate Removal
- Used `duplicated()` to identify exact duplicate records.
- Applied `drop_duplicates()` to ensure duplicate records would not affect the analysis.

### 4. Data Standardization
- Renamed unnamed columns using customer segments and shipping modes.
- Removed redundant subtotal columns to avoid repeated sales information.
- Standardized column names to make the dataset easier to analyze.

### 5. Outlier Detection and Treatment
- Selected the IQR method to identify potential outliers in sales columns.
- Decided to retain all detected outliers because high sales values may represent genuine transactions.

### 6. Data Type Correction
- Converted Order ID to string because it represents an identifier.
- Converted sales columns to `float64` to support numerical calculations.

### 7. Before vs. After Comparison
- Decided to compare missing values, duplicates, row counts, and data types before and after cleaning.
- Used the comparison to evaluate the impact of cleaning decisions.

### 8. Cleaned Dataset Export
- Selected CSV format for the cleaned dataset.
- Used `index=False` to avoid saving unnecessary index values.
- Reloaded the exported file to verify that it was saved successfully.
